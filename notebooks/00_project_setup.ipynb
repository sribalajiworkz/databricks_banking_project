# Banking Lakehouse — Project Setup

This notebook configures and verifies the development environment.

We will:
- Define the project parameters.
- Verify access to the Unity Catalog catalog.
- Prepare Bronze, Silver, and Gold schemas.
- Verify the landing and pipeline state volumes.


```python
dbutils.widgets.dropdown("env","dev",["dev","prod"])
env = dbutils.widgets.get("env")
print(f"Selected Environment: {env}")
```

```python
catalog = "banking_workspace"

bronze_schema= f"bronze_{env}"
silver_schema = f"silver_{env}"
gold_schema = f"gold_{env}"

print(f"Bronze: {catalog}.{bronze_schema}")
print(f"Silver: {catalog}.{silver_schema}")
print(f"Gold:   {catalog}.{gold_schema}")

```

```python
for schema in [bronze_schema, silver_schema, gold_schema]:
    spark.sql(
        f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema}`"
    )
    print(f"Schema ready: {catalog}.{schema}")
```

```python
landing_root = "/Volumes/banking_workspace/files/landing"
state_root = "/Volumes/banking_workspace/files/pipeline_state"

for path in [landing_root,state_root]:
    print(f"\nChecking: {path}")
    for item in dbutils.fs.ls(path):
        print(item.name)
    print("Access Verified")
```

```python
ops_schema = f"ops_{env}"
pipeline_schema = f"pipeline_{env}"

for schema in [ops_schema, pipeline_schema]:
    spark.sql(
        f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema}`"
    )
    print(f"Schema ready: {catalog}.{schema}")
```

```python
landing_path = f"{landing_root}/{env}"
state_path = f"{state_root}/{env}"

dbutils.fs.mkdirs(landing_path)
dbutils.fs.mkdirs(state_path)

print(f"Landing: {landing_path}")
print(f"State:   {state_path}")
```
