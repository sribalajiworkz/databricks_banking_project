```python
display(spark.sql("""
                  SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT customer_id) AS unique_customers,
    COUNT(DISTINCT _source_file) AS source_files
FROM banking_workspace.bronze_dev.customers"""))
```

```python
display(spark.sql("""
SELECT 'customers' AS dataset, COUNT(*) AS total_rows
FROM banking_workspace.bronze_dev.customers
UNION ALL
SELECT 'accounts', COUNT(*)
FROM banking_workspace.bronze_dev.accounts
UNION ALL
SELECT 'branches', COUNT(*)
FROM banking_workspace.bronze_dev.branches
UNION ALL
SELECT 'transactions', COUNT(*)
FROM banking_workspace.bronze_dev.transactions
UNION ALL
SELECT 'merchants', COUNT(*)
FROM banking_workspace.bronze_dev.merchants"""))
```
