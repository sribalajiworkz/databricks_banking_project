```python
display(spark.sql("""
SELECT
    transaction_id,
    COUNT(*) AS active_versions
FROM banking_workspace.silver_dev.transactions_history
WHERE __END_AT IS NULL
GROUP BY transaction_id
HAVING COUNT(*) > 1"""))
```
